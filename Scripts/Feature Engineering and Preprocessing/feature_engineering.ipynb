# Feature Engineering Script

```python
from pathlib import Path
import numpy as np
import pandas as pd

INPUT_PATH = Path('processed_dataset.csv')
OUTPUT_PATH = Path('engineered_dataset.csv')

# If files are loaded in repository layout hide above and unmark below:
# DATA_PATH = Path('/.../.../Datasets/Raw/processed_dataset.csv')
# OUTPUT_PATH = Path('/.../.../Datasets/engineered_dataset.csv')

df = pd.read_csv(INPUT_PATH)

print('Processed shape:', df.shape)
```

```python
# Monetary features
df['discount_amount'] = df['product_price'] * (df['discount_percent'] / 100.0)
df['net_price'] = df['product_price'] - df['discount_amount']

# Engagement features
df['views_per_minute'] = df['num_product_views'] / df['session_length_minutes'].clip(lower=1e-6)
df['price_per_view'] = df['product_price'] / df['num_product_views'].clip(lower=1)

# Customer-history interaction
df['purchase_return_interaction'] = df['past_purchase_count'] * df['past_return_rate']

# Simple interpretable flags
df['high_discount_flag'] = (df['discount_percent'] >= 50).astype(int)
df['delayed_delivery_flag'] = (df['delivery_delay_days'] > 0).astype(int)

new_features = ['discount_amount','net_price','views_per_minute','price_per_view','purchase_return_interaction','high_discount_flag','delayed_delivery_flag']
print('New features:', new_features)
print(df[new_features].describe().T)
```

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(OUTPUT_PATH, index=False)
print(f'Engineered shape: {df.shape}')
print(f'Saved engineered dataset to: {OUTPUT_PATH.resolve()}')
```
