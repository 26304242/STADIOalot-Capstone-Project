# Preprocessing Script

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path('raw_dataset.csv')
OUTPUT_PATH = Path('processed_dataset.csv')

# If files are loaded in repository layout hide above and unmark below:
# DATA_PATH = Path('/.../.../Datasets/Raw/raw_dataset.csv')
# OUTPUT_PATH = Path('/.../.../Datasets/Processed/processed_dataset.csv')

df = pd.read_csv(DATA_PATH)

print('Raw shape:', df.shape)
print('Missing values:', int(df.isna().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))
```

```python
# Removing Exact Duplicats
df = df.drop_duplicates().copy()

# Validating the Target and Identifier
if 'returned' not in df.columns:
    raise ValueError("Target column 'returned' was not found.")
if 'order_id' not in df.columns:
    raise ValueError("Identifier column 'order_id' was not found.")

df['returned'] = df['returned'].astype(int)
print('Processed shape:', df.shape)
print('Target distribution:\n', df['returned'].value_counts())
```

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(OUTPUT_PATH, index=False)
print(f'Saved processed dataset to: {OUTPUT_PATH.resolve()}')
print('Saved columns:', len(df.columns))
```
