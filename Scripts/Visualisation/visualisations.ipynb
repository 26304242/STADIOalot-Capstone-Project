# Visualisations Script

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = Path('engineered_dataset.csv')

# If files are loaded in repository layout hide above and unmark below:
# DATA_PATH = Path('/.../.../Datasets/Engineered/engineered_dataset.csv')

df=pd.read_csv(DATA_PATH)
print(df.shape)
```

```python
df.groupby('returned').size().plot(kind='bar')
plt.title('Target Class Distribution')
plt.xlabel('Returned')
plt.ylabel('Count')
plt.show()

fig, ax = plt.subplots()
sns.boxplot(data=df, x="returned", y="past_return_rate", ax=ax)
ax.set_title("Past Return Rate by Outcome")
plt.show()

fig, ax = plt.subplots()
sns.boxplot(data=df, x="returned", y="discount_percent", ax=ax)
ax.set_title("Discount Percentage by Outcome")
plt.show()
```
